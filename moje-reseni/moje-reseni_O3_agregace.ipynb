```python
# Zjisti min a max délky křídel pro každý druh
vysledek = df.groupby('species')['flipper_length_mm'].agg(['min', 'max'])
vysledek

# Přidej sloupec s rozdílem
vysledek['rozdil'] = vysledek['max'] - vysledek['min']
vysledek

# Najdi druh s nejmenším rozdílem
nejmensi_rozdil = vysledek['rozdil'].min()
druh_s_nejmensim_rozdilem = vysledek['rozdil'].idxmin()
print(f"\nDruh s nejmenším rozdílem: {druh_s_nejmensim_rozdilem}")
print(f"Rozdíl: {nejmensi_rozdil} mm")
```

```python
# Zjisti medián hmotnosti pro každý ostrov
median_hmotnosti = df.groupby('island')['body_mass_g'].median()
print(median_hmotnosti)

# Seřaď sestupně, abys viděl největší medián
median_serazeny = median_hmotnosti.sort_values(ascending=False)
print("\nSeřazeno sestupně:")
print(median_serazeny)

# Najdi ostrov s největším mediánem
ostrov_s_max_medianem = median_hmotnosti.idxmax()
max_median = median_hmotnosti.max()

print(f"\nOstrov s největším mediánem: {ostrov_s_max_medianem}")
print(f"Medián hmotnosti: {max_median} g")
```
